```python
#1. what is unsupervised learning in the context of machine learning?

- Unsupervised learning is a type of machine learning where the model is trained on unlabeled data. the goal is to find hidden structures, patterns, or groupings in the data without prior knowleadge of output labels. Clustering is a common unsupervised learning method.

#2. how does K-Means clustering algorithm work?

-. choose number of cluster k.
 . Randomly initialize k centroids.
 . Assign each data point to the nearest centroid.
 . update centroids by calculating the mean of assigned points.
 . Repeat assignment + update until convergence (centroids don't change significantly).

#3. Explain the concept of a dendrogram in hierarchical clustering.

- A dendrogram is a tree-like diagram that shows hoe clusters are formed step by step in hieararchical clustering. the vertical axis represents distance or dissimilarity, and branches show how points/clusters merge.

#4. what is the main difference between k-means and hierarchical clustering?

. K-Means: partition - based, requires predefinning k, works well for large datasets.

. hierarchical : build a hierarchy of clusters, does not require predefinning clusters but is computationally expensive for large datasets.

#5. what are the advantages of DBSCAN over k-Means?

 . can find clusters of arbitrary shapes.
 . Automatically detects the number of clusters.
 . identifies noise/outliers.
 . No need to specify k in advance.

#6. When would you use silhouette score in clustering?

- when you want to evaluate how ell clusters are separated and how compact they are. Higher silhouette score = better clustering.

#7. what are the limitation of hierarchical clustering?

-. High computational cost for large datasets.
 . once a merge/split is made , it cannot be undone.
 . Sensitive to noise and outliers.

#8. why is feature scaling important in clustering algorithms like k-means?

. because k-means uses distance - based measures (like Euclidean distance). Feature with larger ranges can dominate, so scaling ensures fair contribution of all features.

#9. how does DBSCAN identify noise points?

. points that do not have enough neighbors within the specified radius and cannot be reached from any dense region are marked as noise.

#10. Define interia in the context of k-means.

. Inertia is the sum of squared distances of each point to its assigned centroid. Lower inertia = tighter clusters.

#11. what is the elbow method in k- means clustering?

. A  method to choose the optimal number of clusters(k). Plot inertia vs k and look for the"elbow" point where inertia starts decreasing slowly.

#12. Describe the concept of "density" in DBSCAN.

. Density is defined as the number of points within a given radius. Dense regions form clusters, while sparse regions are noise.

#13. can hierarchical clustering be used on categorical data?

. Yes, but a suitable distance metric must be used instead of Euclidean distance.

#14. What does a negative silhouette Score indicate?

. That the data point may be assigned to the wrong cluster, as it is closer to points in another cluster than its own.

#15. Explain the term "linkage criteria" in hierarchical clustering.

. Linkage criteria determine how distances between clusters are calculated:
. Single linkage(min distance)
.complete linkage (max distance)
.Average linkage (mean distance)

#16. Why might K-Means clustering perform porly on data with varying cluster sizes or densities?

. Because K-Means assumes clusters are spherical and equally sized. It struggles with elongated, overlapping, or uneven density clusters.

#17. What are the core parameters in DBSCAN, and how do they influence clustering?

.eps: Maximum distance between two points to be neighbors.
.min_samples: Minimum number of neighbors required to form a dense region.
These two control cluster shape and noise detection.

#18. How does K-Means++ improve upon standard k-Means initialzation?

.It chooses initial centroids more strategically, spreading them out before clustering. this reduces chances of poor clustering and speeds up convergence.

#19. What is agglomerative clustering?

. A bottom-up hierarchical clustering method where each data point starts as its own cluster, and clusters are merged step by step based on similarity.

#20. What makes Silhouette Score a better metric than just inertia for model evaluation?

. Inertia only measures compactness(lower=better), but not separation.

. Silhouette Score consider both intra-cluster similarity and inter-cluster separation, giving a more reliable evaluation.
```
