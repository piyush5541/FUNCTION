```python
# 1. What are Large Language Models (LLMs) and how do they function?

. Large Language Models (LLms) are deep learning models trained on massive amounts of text data to understand and generate huamn-like language. They are usually based on the Transformer architecture, which uses attention mechanisms to capture relationships between words in a sequence.

LLMs function by:

1. Pre-training on large corpora (books,websites, articles) using self-supervised learning.
2. Learning statistical patterns in language , including, grammer, facts, and context.
3. using a neural network with multiple layers and attention heads to predict the next token in a sequence.
4. Fine-tuning on task - specific datasets (e.g., chat, summarization ,Q&A).

# 2. Discuss the impact of LLMs on traditional software development approaches.

. LLMs are changing traditional software development in serveral ways:

Positive impacts:
. Faster development through AI- assisted coding (e.g., Github copilot).
. Automatic code generation , debugging, and documentation.
. Natural language interfaces replacing complex UI logic.
. Rapid prototying using prompt-based development.

Challenges and changes:

. Shift from rule-based programming to prompt enginerring.
. Need for model monitoring and validation.
. New concerns around correctness, security, amd hallucinations.

# 3. What are the key advantages and limitations of using LLMs in real-world application?

. Advantages:
- Can understand and generate natural language.
- Useful for chatbots, summarization, translation, and coding.
- Reduce manual workload and increase automation.
- Adaptable to many domains via fine-tuning or prompting.
. Limitations:
- Hallucinations(incorrect or fabricated answers).
- High computational and cost requirements.
- Bias present in training data.
- Lack of true reasoning and real-world understanding.
- Data privacy and security concerns.

# 4. Describe how different industries are being transformed by the use of LLMs.Provide example.

. Healthcare: Clinical note summarization, patient chatbots, medical coding.
. Finance: Fraud detection, financial report summarization, customer support bots.
. Education: Personalized tutoring , automated grading, content generation.
. Legal: Contract analysis, legal document summarization, case research.
. E-commerce: Product description generation, recommendation systems, customer support bots.

# 5. Compare and contrast LangChain and Llamalndex. What unique problems does each solve?

   Feature                                            Langchain                                                     Llamalndex
 . Purpose                                         Framework for building LLM-powered application.                Framework for indexing and querying data with LLMs
 . Focus                                           Chains,agents,tools,memory                                     Data ingestion,indexing,retrieval
 . Strength                                        Orechestration of workflows                                    Document-based Q&A and retrieval
```

```python
# 7. Integrate Langchain with a third-party..........?

import requests
from langchain_openai import ChatOpenAI
from langchain.prompts import PromptTemplate

# Call a public weather API (example with Open-Meteo)
url = "https://api.open-meteo.com/v1/forecast?latitude=28.61&longitude=77.21&current_weather=true"
response = requests.get(url).json()
weather = response["current_weather"]

# Prepare LLM
llm = ChatOpenAI(model="gpt-3.5-turbo")

prompt = PromptTemplate(
    input_variables=["temp", "windspeed"],
    template="Explain today’s weather in simple terms. Temperature: {temp}°C, Wind Speed: {windspeed} km/h"
)

text = prompt.format(temp=weather["temperature"], windspeed=weather["windspeed"])
result = llm.predict(text)

print(result)
```

```python
# 8. Create a Llamanlndexn implementation......?

from llama_index.core import VectorStoreIndex, SimpleDirectoryReader

# Load documents from a local folder
reader = SimpleDirectoryReader("./data")
documents = reader.load_data()

# Create an index
index = VectorStoreIndex.from_documents(documents)

# Query the index
query_engine = index.as_query_engine()
response = query_engine.query("What is artificial intelligence?")

print(response)

```

```python
# 9. Demonstrate combining with ......?

from llama_index.core import VectorStoreIndex, SimpleDirectoryReader
from langchain_openai import ChatOpenAI

# Load documents
reader = SimpleDirectoryReader("./data")
documents = reader.load_data()

# Build index
index = VectorStoreIndex.from_documents(documents)
query_engine = index.as_query_engine()

# LLM
llm = ChatOpenAI(model="gpt-3.5-turbo")

# Ask question
question = "Explain machine learning"
retrieved_answer = query_engine.query(question)

final_prompt = f"Based on the following information, answer clearly:\n{retrieved_answer}"
final_answer = llm.predict(final_prompt)

print(final_answer)
```

```python
# 10. A legal firm wants to use AI ......?
from llama_index.core import VectorStoreIndex, SimpleDirectoryReader
from langchain_openai import ChatOpenAI

# Load legal documents
reader = SimpleDirectoryReader("./legal_docs")
documents = reader.load_data()

# Build index
index = VectorStoreIndex.from_documents(documents)
query_engine = index.as_query_engine()

# LLM
llm = ChatOpenAI(model="gpt-3.5-turbo")

# Query for legal information
query = "Summarize the key points of the contract"
retrieved = query_engine.query(query)

prompt = f"Summarize the following legal text in simple terms:\n{retrieved}"
summary = llm.predict(prompt)

print(summary)
```
