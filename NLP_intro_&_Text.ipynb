```python
# 1. What is computational Linguistics and how does it relate to NLP?

- Computational Linguistics is an interdisciplinary field that studies how computers can be used to model, understand, and generate human language using formal rules, algorithms, and statistical models. It combines linguistics, computer science, mathematics, and artifical intellgience.
Natural Language Processing is the the practial application of computational linguistics techniques to build systems such as chatbots, translator, speech assistants, and text analyzers. In short , computational linguistics provides the theoretical foundations, while NLP focuses on building real-world language-processing systems.

# 2. Briefly describe the historical evolution of Natural Language Processing.

- NLP began in the 1950s with rule-based systems such as early machine translation experiments. During the 1960s-1980s, symbolic and grammer-based approaches dominated. In the 1990s, statistical methods using probability models and corpora became popular. In the 2000s, machine learning techniques such as Hidden Markov Models and Conditional Random Fields improved performance. From 2010 onwards, deep learning and neural networks(Word2Vec,RNNs,LSTMs,Transformers, and large language models) revolutionized NLP, enabling highly accurate applications like chatbots, translation systems, and voice assistants.

# 3. List and explain three major use cases of NLP in today's tech industry.

- . Chatbots and virtual Assistant: NLP enables systems like customer support bots and voice assistants (Siri, Alexa) to understand and respond to user queries.
  . Sentiment Analysis: Companies analyze customer reviews and social media posts to detect opinions, emotions, and brand perception.
  . Machine Translation: NLP powers translation systems such as Google Translation That convert text from one language to another automatically.

# 4. What is Text normalization and why is it essential in text processing tasks?

- Text normalization is the process of converting text into a standard and consistent format. It includes lowercasing , removing punctuation, correcting spelling, expanding contractions, and handling numbers of symbols.
  Normalization is essential because it reduces noise, improves consistency, and increase the accuracy of downstream NLP tasks such as tokenization, classification, and information retrieval.

# 5. Compare and contrast stemming and lemmatization with examples.

-  Features                          /  Stemming                             /  Lemmatization
. Definition                         . Reduces words to their root form      . Converts words to their dictionary
                                       by cutting suffixes.                    base form.

. Method                             . Rule-based                            . Uses vocabulary and morphological
                                                                               analysis.

. Accuracy                           . Less accurate                         . More accurate

. Output                             . May produce non-words                 . Produces valid words.


Example

WORD        stemming         Lemmatization

running     run              run

studies     studi            study

better      better           good


```

```python
# 6. python program to extract email addresses using Regex.

import re

text = """
Hello team, please contact us at support@xyz.com for technical issues, or reach out to our HR at hr@xyz.org and jenny via jenny_clarke126@mail.co.us. For partnership inquiries, email partners@xyz.biz.
"""
# Corrected pattern: Changed 'O-9' to '0-9' for digits
pattern = r'[a-zA-Z0-9._%+-]+@[a-zA-Z0-9.-]+\.[a-zA-Z]+'

emails = re.findall(pattern, text)

print("Extracted Emails:")
for email in emails:
    print(email)
```

```python
# 7. Tokenization and Frequency Distribution using NLTK

import nltk
# Download the 'punkt' tokenizer models if not already downloaded
nltk.download('punkt')
nltk.download('punkt_tab') # Download 'punkt_tab' as suggested by the error
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist

text = """Natural Language Processing (NLP) is a fascinating field that combines linguistics,
computer science, and artificial intelligence. It enables machines to understand,
interpret, and generate human language."""

tokens = word_tokenize(text)

freq = FreqDist(tokens)

print("Tokens:")
print(tokens)

print("\nFrequency Distribution:")
print(freq.most_common(10))
```

```python
# 8. Custom Annotator to Identify Proper Nouns.

import spacy

nlp = spacy.load("en_core_web_sm")

text = "John works at Google in California with Sarah."

doc = nlp(text)

print("Proper Nouns:")

for token in doc:
    if token.pos_ == "PROPN":
        print(token.text)
```

```python
# 9. Train Word2Vec MOdel using Gensim.

!pip install gensim
from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize

dataset = [
"Natural language processing enables computers to understand human language",
"Word embeddings are a type of word representation that allows words with similar meaning to have similar representation",
"Word2Vec is a popular word embedding technique used in many NLP applications",
"Text preprocessing is a critical step before training word embeddings",
"Tokenization and normalization help clean raw text for modeling"
]

tokenized_data = [word_tokenize(sentence.lower()) for sentence in dataset]

model = Word2Vec(sentences=tokenized_data, vector_size=100, window=5, min_count=1, workers=4)

print(model.wv.most_similar('word'))
```
