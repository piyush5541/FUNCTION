```python
# 1. Compare and contrast NLTK and spaCy in terms of features , ease of use , and performance.

- NLTK (Natural Language Toolkit) and spaCy are both popular libraries for natural language Processing, but they serve slightly different purposes.

. Features:
- NLTK provides a very wide range of tools for tokenization, stemming, lemmatization, parshing, tagging,and access to many linguistic corpora. it is excellent for teaching and research.
- spaCy focuses more on industrial-strength NLP with built-in models for tokenization, POS tagging, dependency parsing, and named entity recognition, but with fewer experimental tools.
. Ease of Use:
- NLTK often requires more setup, downloading corpora, and writing multiple lines of code for simple tasks.
- spaCy is easier and cleaner to use , with ready-made pipelines and simple APIs.
. Performance:
- NLTK is slower and not optimized for large-scale production use.
- spaCy is highly optimized in Cython and is much faster, making it suitable for real-time and production systems.

# 2. What is TextBlob and how does it simplify common NLP tasks like sentiment analysis and translation?

- TextBlob is a Python library built on top of NLTK and pattern that provides a simple API for common NLP tasks. It simplifies tasks such as:
. Sentiment analysis(polarity and subjectivity scoring)
. Part-of-speech tagging
. Noun phrase extraction
. Translation and language detection

TextBlob is easy to use because complex NLP operations can be done with only one or two lines of code. for example, sentiment analysis can be performed by simply calling TextBlob(text).sentiments,
without manually training or loading models.

# 3. Explain the role of stanford NLP in academic and industry NLP prjects.

. Stanford NLP (now known as Stanford CoreNLP and Stanza) plays as important role in both academic research and industry applications.

In academics, it provides state-of the-art models for syntactic parshing, named entity recognition, sentiment analysis, and dependency parsing, which are widely used in research papers and experiments.

In industry , Stanford NLP is used in applications such as:
. Information extraction sysytems
. Question answering systems
. Chatbots and virtual assistants
. Text analytics platforms

# 4. Describe the architecture and functioning of a Recurrent Neural Network (RNN).

- A Recurrent Neural Network is a type of neural network designed to process sequential data such as text, speech, or time-series data.

Architecture:
. Input layer recieves one element of the sequence at a time.
. Hidden layer has recurrent connection, meaning the output from the previous time step is fed back into the network.
. Output layer produces predicitons for each time step or the final step.

Functioning: AT eaach time step , the RNN takes the current input and the previous hidden state to compute a new hidden state. This allows the network to maintain a memory of past input and model sequential dependencies.

# 5. What is the key difference between LGTM and GRU networks in NLP applications?

- The key difference between LGTM and GRU networks lies in their internal gate structures.
. LSTM uses three gates: input gate, forget gate, and output gate, along with a separate memory cell to control long-term information flow.
. GRU combines the input and forget gates into a single update gate and has fewer parameters.

LSTMs are more powerful for capturing very long-term dependencies, while GRUs are simpler, faster to train, and oftenperform similarly with less computational cost.




```

```python
# 6. Sentiment Analysis using TextBlob?

from textblob import TextBlob

text="I had a great experience using the new mobile banking app. The interface is intuitive, and customer support was quick to resolve my issue. However, the app did crash once during a transaction, which was frusting"

blob = TextBlob(text)
sentiment = blob.sentiment

print("Polarity:", sentiment.polarity)
print("Subjectivity:", sentiment.subjectivity)
```

```python
# 7. Tokenization and Frequency Distribution using NLYK?

import nltk
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist

nltk.download('punkt_tab')

text = "Natural Language Processing (NLP) is a fascinating field that combines linguistic, computer science, and artificial intelligence to enable computers to understand, interpret, and generate human language."

tokens = word_tokenize(text.lower())
freq_dist = FreqDist(tokens)

print("Tokens:", tokens[:20])
print("Most common words:")
print(freq_dist.most_common(10))
```

```python
# 8. Basic LSTM Model in Keras for Text Classification?

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
import numpy as np

texts = [
    "I love this project",
    "This is an amazing experience",
    "I hate waiting in line",
    "This is the worst service",
    "Absolutely fantastic!"
]

labels = [1, 1, 0, 0, 1]

# Tokenization
tokenizer = Tokenizer()
tokenizer.fit_on_texts(texts)
sequences = tokenizer.texts_to_sequences(texts)

# Padding
max_len = 6
X = pad_sequences(sequences, maxlen=max_len)
y = np.array(labels)

# Build model
model = Sequential()
model.add(Embedding(input_dim=50, output_dim=16, input_length=max_len))
model.add(LSTM(32))
model.add(Dense(1, activation='sigmoid'))

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Train model
model.fit(X, y, epochs=10, verbose=1)

```

```python
# 9. spaCy NLP Pipeline with Tokenization, Lemmatization, and NER?

import spacy

nlp = spacy.load("en_core_web_sm")

text = "Homi Jehangir Bhaba was an Indian nuclear physicist who played a key role in the development of India’s atomic energy program. He was the founding director of the Tata Institute of Fundamental Research (TIFR) and was instrumental in establishing the Atomic Energy Commission of India."

doc = nlp(text)

print("Tokens and Lemmas:")
for token in doc:
    print(token.text, "->", token.lemma_)

print("\nNamed Entities:")
for ent in doc.ents:
    print(ent.text, "->", ent.label_)
```
